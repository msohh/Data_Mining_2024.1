```python
df = df.drop(columns=['상품URL', '모델번호','색상'])
```

```python
invalid_rows = df[df['발매일로부터현재까지일'].isnull() | df['발매가'].isnull()].index

# 유효한 '발매일로부터현재까지일' 또는 '발매가' 값이 없는 행 삭제
df_cleaned = df.drop(invalid_rows)

# 결과 확인
print(f'Number of rows after removing invalid "발매일로부터현재까지일" and "발매가" rows: {df_cleaned.shape[0]}')
```

```python
# '가격변동성' 컬럼 추가
df_cleaned['가격변동성'] = (100*(df_cleaned['최고거래가'] - df_cleaned['최저거래가']) / df_cleaned['최고거래가']).round(2)

# '수익률' 컬럼 추가
df_cleaned['수익률'] = (100*(df_cleaned['최고거래가'] - df_cleaned['발매가']) / df_cleaned['발매가']).round(2)
```

```python
df_cleaned['콜라보'] = df_cleaned['상품명'].apply(lambda x: 1 if 'x' in x else 0)
```

```python
brand_counts = df_cleaned['브랜드'].value_counts()

# 결과 확인
print('브랜드별 개수:')
print(brand_counts)
```

```python
brands_to_keep = ['Nike', 'Jordan', 'New Balance', 'Adidas', 'Asics', 'Vans', 'Converse', 'Salomon', 'Mihara Yasuhiro', 'Puma']

# 주어진 브랜드에 속하는 행만 필터링
df_filtered = df_cleaned[df_cleaned['브랜드'].isin(brands_to_keep)]
```

```python
from scipy.stats import yeojohnson

columns_to_transform = ['거래량', '발매일로부터현재까지일', '관심', '스타일', '수익률', '최저거래가', '최고거래가', '최근거래가', '발매가', '가격변동성']

# Yeo-Johnson 변환 함수 정의
def yeojohnson_transform(column):
    transformed_values, lambda_value = yeojohnson(column)
    return transformed_values, lambda_value

# Yeo-Johnson 변환 적용
for column in columns_to_transform:
    df_filtered[column], _ = yeojohnson_transform(df_filtered[column])
```

```python
import numpy as np
from scipy.stats import zscore

# # Z-score를 이용한 이상치 제거 함수 정의
def remove_outliers(df, threshold = 1.92):
    numeric_cols = df.select_dtypes(include=[np.number]).columns
    z_scores = df[numeric_cols].apply(zscore)  # 각 컬럼별로 Z-score 계산
    abs_z_scores = np.abs(z_scores)  # Z-score의 절대값 구하기
    filtered_entries = (abs_z_scores < threshold).all(axis=1)  # 각 컬럼별로 threshold 미만인 데이터 유지
    return df[filtered_entries]

# # 이상치 제거
df_filtered = remove_outliers(df_filtered)

# # 결과 확인
print(df_filtered.describe())
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler, Normalizer

# '상품명'을 제외한 컬럼 선택
columns_to_scale = df_filtered.columns.difference(['상품명','발매일','최고거래성사일','최저거래성사일'])

# 각 스케일링 기법을 리스트에 저장
scalers = [
    StandardScaler(),
    MinMaxScaler(),
    RobustScaler(),
]

# 각 스케일링을 적용한 데이터를 저장할 딕셔너리
scaled_data = {}

# 각 스케일링을 적용하고 데이터를 딕셔너리에 저장
for scaler in scalers:
    if isinstance(scaler, Normalizer) or isinstance(scaler, np.ufunc):
        scaled_data[str(scaler)] = scaler(df_filtered[columns_to_scale].values)  # 로그 변환 및 정규화는 fit_transform 메서드를 사용하지 않음
    else:
        scaled_data[str(scaler)] = scaler.fit_transform(df_filtered[columns_to_scale])

# 결과 확인
for scaler, scaled_df in scaled_data.items():
    print(f"Scaler: {scaler}\n{scaled_df}\n")
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, MeanShift, DBSCAN

# 사용할 클러스터링 알고리즘들
clustering_algorithms = {
    "KMeans": KMeans(n_clusters=3, random_state=42),
    "MeanShift": MeanShift(),
    "DBSCAN": DBSCAN(eps=0.5, min_samples=5)
}

# StandardScaler를 사용하여 스케일링
scaler = StandardScaler()
scaled_features = scaler.fit_transform(df_filtered[['수익률', '발매일로부터현재까지일', '콜라보']])

# 각 클러스터링 알고리즘을 순회하면서 적용
for name, algorithm in clustering_algorithms.items():
    algorithm.fit(scaled_features)
    df_filtered[name+"_cluster"] = algorithm.labels_
    print(f"Clusters using {name}:")
    print(df_filtered[['수익률', '발매일로부터현재까지일', '콜라보', name+"_cluster"]])
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# 클러스터 개수 후보
cluster_range = range(1, 11)
inertia = []

# 각 클러스터 개수에 대한 KMeans 모델 생성 후 inertia 측정
for num_clusters in cluster_range:
    kmeans = KMeans(n_clusters=num_clusters, random_state=42)
    kmeans.fit(df_filtered[['수익률', '발매일로부터현재까지일', '콜라보']])
    inertia.append(kmeans.inertia_)

# 엘보우 메서드 시각화
plt.plot(cluster_range, inertia, marker='o', linestyle='-')
plt.xlabel('Number of clusters')
plt.ylabel('Inertia')
plt.title('Elbow Method')
plt.show()
```
