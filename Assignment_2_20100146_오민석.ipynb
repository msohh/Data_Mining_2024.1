```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_digits
digits = load_digits()
```

Data load, 데이터 파악

```python
print("Feature names:\n", digits['feature_names'])
print("Target names:", digits['target_names'])
```

```python
print(digits['data'].shape)
print(digits['target'].shape)
```

```python
df = pd.DataFrame(digits.data, columns = digits.feature_names)
df['target'] = digits.target
df
```

```python
from sklearn.model_selection import train_test_split
X_train1, X_test, y_train1, y_test = train_test_split(digits.data, digits.target, random_state = 0)
X_train, X_val, y_train, y_val = train_test_split(X_train1, y_train1, random_state = 0)
```

```python
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_val shape:", X_val.shape)
print("y_val shape:", y_val.shape)
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

데이터 전처리

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler.fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)
```

Decision Tree Classifier

Hyperparmeter: max_depth, min_samples_leag, max_leaf_nodes

```python
training_accuracy = []
val_accuracy = []
max_depth = []
min_samples_leaf = []
max_leaf_nodes = []

md_settings = [10, 100, 1000]
msl_settings = [1, 10, 100]
mln_settings = [10, 100, 1000]

for md in md_settings:
    for msl in msl_settings:
        for mln in mln_settings:
        # model training with labeled training data
            dt = DecisionTreeClassifier(max_depth = md, min_samples_leaf = msl, max_leaf_nodes= mln, random_state=0)
            dt.fit(X_train, y_train)

        # prediction
            y_train_hat = dt.predict(X_train)
            y_val_hat = dt.predict(X_val)

        # evaluation    
            training_accuracy.append(accuracy_score(y_train, y_train_hat))
            val_accuracy.append(accuracy_score(y_val, y_val_hat))
            max_depth.append(md)
            min_samples_leaf.append(msl)
            max_leaf_nodes.append(mln)

result_dt = pd.DataFrame({"max_depth": max_depth, "min_samples_leaf": min_samples_leaf, "max_leaf_nodes": max_leaf_nodes, "training accuracy": training_accuracy, "val accuracy": val_accuracy})
result_dt
```

SVM

Hyperparmeter: C, gamma

```python
training_accuracy = []
val_accuracy = []

C_settings = [0.01, 0.1, 1, 10, 100]
gamma_settings = [0.01, 0.1, 1, 10, 100]
for C in C_settings:
    for gamma in gamma_settings:
        # model training with labeled training data
        svc = SVC(C = C, gamma = gamma, random_state = 1)
        svc.fit(X_train_scaled, y_train)

        # prediction
        y_train_hat = svc.predict(X_train_scaled)
        y_val_hat = svc.predict(X_val_scaled)

        # evaluation    
        training_accuracy.append(accuracy_score(y_train, y_train_hat))
        val_accuracy.append(accuracy_score(y_val, y_val_hat))

result_svm = pd.DataFrame({"C":sorted(C_settings*5), "gamma":gamma_settings*5, "training accuracy": training_accuracy, "validation accuracy": val_accuracy})
result_svm
```

RandomForesrClassifier

Hyperparmeter: n_estimators

```python
training_accuracy = []
val_accuracy = []

n_settings = [1, 5, 10, 15, 50, 100]

for n in n_settings:
    # model training with labeled training data
    rf = RandomForestClassifier(n_estimators = n, random_state = 2)
    rf.fit(X_train_scaled, y_train)

    # prediction
    y_train_hat = rf.predict(X_train_scaled)
    y_val_hat = rf.predict(X_val_scaled)

    # evaluation
    training_accuracy.append(accuracy_score(y_train, y_train_hat))
    val_accuracy.append(accuracy_score(y_val, y_val_hat))

pd.DataFrame({"n_estimators": n_settings, "training accuracy": training_accuracy, "validation accuracy": val_accuracy})
```

최적의 learning algorithm 선택
hyperparameter
DecisionTree: max_depth = 10, min_samples_leaf = 1, max_leaf_nodes= 100
SVM: C = 10, gamma = 0.01
RandomForest: n_estimators = 50

```python
dt = DecisionTreeClassifier(max_depth = 10, min_samples_leaf = 1, max_leaf_nodes= 100)
svc = SVC(C = 10, gamma = 0.01)
rf = RandomForestClassifier(n_estimators = 50)

dt.fit(X_train, y_train)
svc.fit(X_train_scaled, y_train)
rf.fit(X_train_scaled, y_train)

y_test_hat_dt = dt.predict(X_test)
y_test_hat_svc = svc.predict(X_test_scaled)
y_test_hat_rf = rf.predict(X_test_scaled)

print("evaluation of Decision Tree: {:.2f}" .format(accuracy_score(y_test, y_test_hat_dt)))
print("evaluation of SVM: {:.2f}" .format(accuracy_score(y_test, y_test_hat_svc)))
print("evaluation of Random Forest: {:.2f}" .format(accuracy_score(y_test, y_test_hat_rf)))
```
